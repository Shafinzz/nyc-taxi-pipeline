```python
# Path to ONE month of data (January 2019)
path = "/databricks-datasets/nyctaxi/tripdata/yellow/yellow_tripdata_2019-01.csv.gz"

# Ask Spark to read that CSV file (first row = column names)
df = spark.read.csv(path, header=True)

# Show the first 1,000 rows as a table
display(df.limit(1000))
```

```python
# 1) Show every column name and its data type
df.printSchema()

# 2) Count how many rows (rides) are in January 2019
print("Number of rides in January 2019:", df.count())
```

```python
# Sort trip_distance from largest to smallest, while it's still TEXT
df.select("trip_distance").orderBy("trip_distance", ascending=False).show(5)
```

```python
from pyspark.sql import functions as F

# Convert trip_distance from text to a real number, then sort biggest first
(df.select(F.col("trip_distance").cast("double").alias("distance_miles"))
   .orderBy("distance_miles", ascending=False)
   .show(5))
```

```python
from pyspark.sql import functions as F

# Add a new column: trip distance as a real number
df_num = df.withColumn("distance_miles", F.col("trip_distance").cast("double"))

# Count trips that are too long, and trips with zero (or negative) distance
too_long = df_num.filter(F.col("distance_miles") > 300).count()
zero_or_less = df_num.filter(F.col("distance_miles") <= 0).count()

print("Trips over 300 miles:", too_long)
print("Trips with 0 or negative miles:", zero_or_less)
```

```python
from pyspark.sql import functions as F

# Convert the pickup time from text to a real timestamp, then take the hour (0–23)
rides_by_hour = (
    df.withColumn("pickup_hour", F.hour(F.to_timestamp("tpep_pickup_datetime")))
      .groupBy("pickup_hour")
      .count()
      .orderBy("pickup_hour")
)

display(rides_by_hour)
```

```python
# Read January, but keep only 1,000 rows (small, for practice)
path = f"{SOURCE_PATH}/yellow_tripdata_2019-01.csv.gz"
sample = spark.read.csv(path, header=True).limit(1000)

# Save those rows as a Delta table called practice_trips
sample.write.format("delta").mode("overwrite").saveAsTable(f"{CATALOG}.{SCHEMA}.practice_trips")

print("✅ Saved practice_trips")
```

```python
TABLE = f"{CATALOG}.{SCHEMA}.practice_trips"   # short name for our practice table

display(spark.sql(f"DESCRIBE HISTORY {TABLE}"))
```

```python
# Take 500 rows and ADD them to the table (don't replace it)
more = spark.read.csv(path, header=True).limit(500)
more.write.format("delta").mode("append").saveAsTable(TABLE)

print("Rows now:", spark.table(TABLE).count())
```

```python
display(spark.sql(f"DESCRIBE HISTORY {TABLE}"))
```

```python
# How many rows did the table have at version 1 (before we added 500)?
old = spark.sql(f"SELECT * FROM {TABLE} VERSION AS OF 1")
print("Rows at version 1:", old.count())

# How many rows does it have now?
print("Rows now:", spark.table(TABLE).count())
```

```python
# OOPS! Delete every row in the table (on purpose, practice table only!)
spark.sql(f"DELETE FROM {TABLE}")

print("Rows after the accident:", spark.table(TABLE).count())
```

```python
# Bring the table back to how it was at version 2 (before the accident)
spark.sql(f"RESTORE TABLE {TABLE} TO VERSION AS OF 2")

print("Rows after restore:", spark.table(TABLE).count())
```

```python
from pyspark.sql import functions as F
TABLE = f"{CATALOG}.{SCHEMA}.practice_trips"
```

```python
# Take 10 rows and add a NEW column called "note"
extra = spark.table(TABLE).limit(10).withColumn("note", F.lit("hello"))

# Try to append them to the table
extra.write.format("delta").mode("append").saveAsTable(TABLE)
```

```python
# Same thing, but this time we ALLOW the new column
extra.write.format("delta").mode("append").option("mergeSchema", "true").saveAsTable(TABLE)

# Check: what columns does the table have now?
print(spark.table(TABLE).columns)

# How many rows have a value in the new "note" column?
print("Rows with a note:", spark.table(TABLE).filter(F.col("note").isNotNull()).count())
```

```python
# Remove the practice table (we don't need it anymore)
spark.sql(f"DROP TABLE IF EXISTS {TABLE}")
print("🧹 practice_trips removed")
```
