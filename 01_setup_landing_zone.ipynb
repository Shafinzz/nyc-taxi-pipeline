# 01 · Setup landing zone
**Purpose:** Simulate taxi files "arriving" by copying the first 3 months (Jan–Mar 2019) into our landing folder.
**Why:** The pipeline must process new files incrementally. April 2019 is held back and added later to prove only new data gets processed (REQ-05).
**Output:** 3 CSV files in `/Volumes/workspace/nyc_taxi/pipeline_files/landing`

```python
%run ./00_config
```

```python
# Create the schema and volume if they don't exist yet (safe to re-run)
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOG}.{SCHEMA}")
spark.sql(f"CREATE VOLUME IF NOT EXISTS {CATALOG}.{SCHEMA}.{VOLUME}")
print("✅ Schema and volume ready")
```

```python
# Create the landing folder (if it already exists, nothing happens)
dbutils.fs.mkdirs(LANDING_PATH)

# Copy each starting month from the shared dataset into OUR landing folder
for month in START_MONTHS:
    file_name = f"yellow_tripdata_{month}.csv.gz"   # e.g. yellow_tripdata_2019-01.csv.gz
    source    = f"{SOURCE_PATH}/{file_name}"         # where it comes from
    target    = f"{LANDING_PATH}/{file_name}"        # where it goes
    dbutils.fs.cp(source, target)
    print("Copied:", file_name)
```

```python
# List the files now in our landing folder, with their size in MB
for f in dbutils.fs.ls(LANDING_PATH):
    print(f.name, "-", round(f.size / 1_000_000, 1), "MB")
```
